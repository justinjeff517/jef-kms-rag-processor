```python
import boto3
from botocore.client import Config
from dotenv import load_dotenv
load_dotenv()
import os
# DigitalOcean Spaces credentials
ACCESS_KEY = os.getenv('SPACES_ACCESS_KEY')
SECRET_KEY = os.getenv('SPACES_SECRET_KEY')
SPACE_NAME = 'jef-bucket-one'
REGION = 'sgp1'  # Example: 'nyc3'
SPACE_URL = f'https://{SPACE_NAME}.{REGION}.digitaloceanspaces.com'

# List of keys (file names in DigitalOcean Spaces)
pdf_keys = [
    'kms/civil-engineering/college-algebra-by-barnett-ninth-edition/100.pdf',
    'kms/civil-engineering/college-algebra-by-barnett-ninth-edition/110.pdf',

    'kms/civil-engineering/college-algebra-by-barnett-ninth-edition/120.pdf',


]

# Connect to DigitalOcean Spaces
session = boto3.session.Session()
client = session.client('s3',
                        region_name=REGION,
                        endpoint_url=f'https://{REGION}.digitaloceanspaces.com',
                        aws_access_key_id=ACCESS_KEY,
                        aws_secret_access_key=SECRET_KEY,
                        config=Config(signature_version='s3v4'))

# Generate public URLs
for key in pdf_keys:
    # If the files are private, you can generate pre-signed URLs
    url = client.generate_presigned_url('get_object',
                                        Params={'Bucket': SPACE_NAME, 'Key': key},
                                        ExpiresIn=3600)  # URL valid for 1 hour
    print(f'Access URL for {key}: {url}')

    # If the files are public, you can directly construct the URL:
    # public_url = f'{SPACE_URL}/{key}'
    # print(f'Public URL for {key}: {public_url}')
```
