```python
import requests
from requests.auth import HTTPBasicAuth
from dotenv import load_dotenv
import os

# Load environment variables
load_dotenv()

# Retrieve credentials and endpoint
spaces_endpoint = "sgp1.digitaloceanspaces.com"  # Use the base region endpoint
spaces_access_key = os.getenv("SPACES_ACCESS_KEY")
spaces_secret_key = os.getenv("SPACES_SECRET_KEY")
bucket_name = "jef-bucket-one"  # Your bucket name

# Construct the virtual-hosted style URL
url = f"https://{bucket_name}.{spaces_endpoint}/?delimiter=/"

# Set headers for S3 compatibility
headers = {
    "Host": f"{bucket_name}.{spaces_endpoint}"
}

# Make the GET request to list folders
response = requests.get(url, auth=HTTPBasicAuth(spaces_access_key, spaces_secret_key), headers=headers)

# Check if the request was successful
if response.status_code == 200:
    from xml.etree import ElementTree as ET
    root = ET.fromstring(response.content)
    
    folders = root.findall(".//CommonPrefixes/Prefix")
    if folders:
        print("Folders in the bucket:")
        for folder in folders:
            print(f"- {folder.text}")
    else:
        print("No folders found in the bucket.")
else:
    print(f"Error listing folders: {response.status_code} - {response.text}")
```
