```python
import os
from pypdf import PdfReader, PdfWriter

def split_pdf_by_page(input_pdf, output_folder="outputs"):
    # Create the output folder if it doesn't exist
    os.makedirs(output_folder, exist_ok=True)
    
    # Load the input PDF
    reader = PdfReader(input_pdf)
    
    # Iterate over all pages
    for i, page in enumerate(reader.pages, start=1):  # start=1 for 1-based indexing
        # Create a new PDF writer for each page
        writer = PdfWriter()
        writer.add_page(page)
        
        # Define the output filename (only the page number)
        output_filename = os.path.join(output_folder, f"{i}.pdf")
        
        # Write the single-page PDF
        with open(output_filename, "wb") as output_pdf:
            writer.write(output_pdf)
        
        print(f"Page {i} saved as: {output_filename}")

# Usage
input_pdf = "sample.pdf"  # Replace with your actual PDF file
split_pdf_by_page(input_pdf)
```
