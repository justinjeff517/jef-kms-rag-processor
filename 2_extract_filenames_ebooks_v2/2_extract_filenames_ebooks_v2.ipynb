```python
import os
import json

# Specify the target folder
folder_path = r"E:\ebooks_v2"

# List to store PDF filenames only (without paths)
pdf_files = []

# Walk through all subdirectories
for root, _, files in os.walk(folder_path):
    for file in files:
        if file.lower().endswith(".pdf"):
            pdf_files.append(file)  # Only store the filename, not the path

# Define output JSON file path inside the same folder
output_file = os.path.join("ebooks_v2_pdf_files.json")

# Save the PDF filenames to a JSON file in the same folder
with open(output_file, "w", encoding="utf-8") as f:
    json.dump(pdf_files, f, indent=4)

# Print confirmation
print(f"PDF filenames saved to: {output_file}")
```
